TRANSFORMATION : I will make a copy of the dataset,  I will drop the 2 leakage columns(Naive_Bayes_Classifier), create the churn column, Encode categories, Handle Outliers 

```python
# Making a copy of the original data frame as the  working copy for cleaning and transformation to preserve the original data frame for reference.
df_clean = df.copy()
df_clean.shape
```

```python
# Dropping the columns that are not useful for analysis and modeling.
columns_to_drop = ['Naive_Bayes_Classifier_Attrition_Flag_Card_Category_Contacts_Count_12_mon_Dependent_count_Education_Level_Months_Inactive_12_mon_1', 'Naive_Bayes_Classifier_Attrition_Flag_Card_Category_Contacts_Count_12_mon_Dependent_count_Education_Level_Months_Inactive_12_mon_2']
df_clean.drop(columns=columns_to_drop, inplace=True)

```

```python
df_clean.shape
```

```python
# Creating a new column to indicate whether a customer is a churner or not based on the 'Attrition_Flag' column.
df_clean["Churn_Flag"] = df_clean["Attrition_Flag"].map({
    "Attrited Customer": 1,
    "Existing Customer": 0
})  
```

```python
df_clean["Churn_Flag"].value_counts()
df_clean["Churn_Flag"].mean()
```

The attrition label was transformed into binary churn indicator using encoding where churned customer were assigned 1 and existing customer a value of 0. The Churn rate is 16.1% . Out of the  10,127 customers: 16.1% have churned while 83.9% are retained. But the churn rate is still a significant revenue risk. 

```python
# Creating Tenure Bands 
df_clean["Tenure_Band"] = pd.cut(df_clean["Months_on_book"], bins=[0, 12, 24, 36, 48, 60], labels=["0-1 year", "1-2 years", "2-3 years", "3-4 years", "4+ years"], include_lowest=True)
```

```python
df_clean["Tenure_Band"].value_counts()
```

```python
# previewing the mapping 
df_clean[["Months_on_book", "Tenure_Band"]].head(10)
```

```python
# Creating Age Bands
df_clean["Age_Band"] = pd.cut(df_clean["Customer_Age"], bins=[18, 30, 40, 50, 60, 100], labels=["18-30", "31-40", "41-50", "51-60", "60+"], include_lowest=True)
df_clean[["Customer_Age", "Age_Band"]].head(10)
```

```python
# Creating Utilisation Segments 
df_clean["Utilisation_Segment"] = pd.cut(df_clean["Avg_Utilization_Ratio"], bins=[0, 0.3, 0.7, 1.0], labels=["Low", "Medium", "High",], include_lowest=True)
```

```python
df_clean[["Avg_Utilization_Ratio", "Utilisation_Segment"]].head(10)
df_clean["Utilisation_Segment"].value_counts()
```

```python
# Creating Activity_Level Segments
df_clean["Activity_Level"] = pd.cut(df_clean["Total_Trans_Ct"], bins=[0, 40, 80, 200], labels=["Low", "Medium", "High"], include_lowest=True)
df_clean[["Total_Trans_Ct", "Activity_Level"]].head(10)
df_clean["Activity_Level"].value_counts()
```

```python
# Creating spend_Drop flag
df_clean["Spend_Drop_Flag"] = (df_clean["Total_Amt_Chng_Q4_Q1"] < 1).astype(int )
df_clean[["Total_Amt_Chng_Q4_Q1", "Spend_Drop_Flag"]].head(10)
df_clean["Spend_Drop_Flag"].value_counts()
```

```python
# Creating Transaction _Drop flag
df_clean["Transaction_Drop_Flag"] = (df_clean["Total_Ct_Chng_Q4_Q1"] < 1).astype(int )
df_clean[["Total_Ct_Chng_Q4_Q1", "Transaction_Drop_Flag"]].head(10)
df_clean["Transaction_Drop_Flag"].value_counts()    
```

```python
# saving the cleaned and transformed data frame to a new csv file for analysis and modeling.
df_clean.to_csv("../DataSet/Cleaned/BankChurners_Cleaned.csv", index=False)     
```
